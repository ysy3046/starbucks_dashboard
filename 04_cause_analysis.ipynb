# 원인 분해 초안

## 1. 초기 지표에서 확인된 문제

관찰기간이 충분히 확보된 프로모션 수신 건을 기준으로 분석한 결과:

- 전체 수신 후 열람률: 74.83%
- BOGO·할인형 열람 후 완료율: 50.69%
- BOGO 열람 후 완료율: 44.15%
- 할인형 열람 후 완료율: 60.19%

BOGO는 수신 후 열람률이 83.77%로 높았지만, 열람 후 완료율은 할인형보다 16.04%p 낮았다.

따라서 다음 문제를 우선 분석 대상으로 설정한다.

> BOGO는 고객의 열람을 유도하지만, 열람 이후 프로모션 완료로 충분히 이어지지 않는다.

---

## 2. 원인 분해

### 문제: BOGO의 열람 후 완료율이 낮다

#### 1) 프로모션별 차이

BOGO 4개의 열람 후 완료율은 38.27%~52.48%로 차이가 나타났다.

- 프로모션 ID별로 완료율 차이가 있는가?
- 성과가 낮은 특정 BOGO가 전체 완료율을 낮추고 있는가?

#### 2) 달성 조건

난이도가 높은 프로모션의 열람 후 완료율은 48.18%로, 난이도가 낮은 프로모션의 54.14%보다 낮았다.

- 최소 결제 조건이 높아 완료하지 못하는가?
- 고객의 평소 구매 수준보다 달성 조건이 높은가?

#### 3) 보상 구조

보상이 평균보다 높은 프로모션의 열람 후 완료율은 44.87%로, 평균 이하 프로모션의 60.94%보다 낮았다.

- 높은 보상이 높은 완료율로 이어지지 않는가?
- 보상 자체보다 달성 조건과 보상의 균형이 중요한가?

#### 4) 유효기간

유효기간이 긴 프로모션의 열람 후 완료율은 54.72%로, 짧은 프로모션의 43.70%보다 높았다.

- 유효기간이 짧아 고객이 조건을 달성할 시간이 부족한가?

#### 5) 채널 조합

mobile·social이 포함된 프로모션은 열람률이 높았지만 열람 후 완료율은 더 높지 않았다.

- 채널 조합은 프로모션 열람에는 연관되지만 완료에는 직접 연결되지 않는가?
- 노출 확대만으로는 완료율을 높이기 어려운가?

#### 6) 고객 특성

- 여성의 열람 후 완료율이 남성보다 높았다.
- 소득 8만 달러 이상 고객의 완료율이 더 높았다.
- 55세 이상 고객의 완료율이 54세 이하보다 높았다.
- 고객정보 미입력 집단의 완료율은 매우 낮았다.

고객 구성의 차이가 프로모션별 완료율 차이에 영향을 주었는지 확인할 필요가 있다.

---

## 3. 행동 원인 가설

BOGO는 혜택에 대한 관심을 끌어 열람까지는 유도하지만, 할인형보다 추가적인 구매 계획이나 조건 달성이 필요해 완료로 이어지지 않을 가능성이 있다.

다만 고객이 프로모션을 완료하지 않은 실제 이유는 현재 데이터에 없다. 따라서 망각, 구매 필요성 부족, 조건의 불편함 등은 현재 데이터로 확정할 수 없는 가설로 남긴다.

---

## 4. 추가 검증이 필요한 항목

- BOGO 4개 사이의 완료율 차이
- 난이도와 완료율의 관계
- 유효기간과 완료율의 관계
- 고객 특성을 통제한 후에도 BOGO와 할인형의 차이가 유지되는지
- 고객정보 미입력 집단의 완료율이 낮은 이유
- 각 집단의 차이가 통계적으로 유의한지

---

## 5. 현재 단계의 결론

현재 분석에서는 BOGO의 높은 열람률이 높은 완료율로 이어지지 않는 문제가 확인되었다. 프로모션별 달성 조건, 유효기간과 고객 구성이 완료율 차이와 연관되어 있을 가능성이 있다.

이 내용은 기술통계에 따른 원인 분해 초안이며, 팀 논의와 통계 검정을 거쳐 최종 원인 구조를 확정한다.
