# 문제 정의 초안

## 1. 대시보드 사용자

스타벅스 프로모션을 기획하고 운영하는 마케팅 담당자

## 2. 문제 상황

프로모션의 전체 완료율만 확인하면 성과가 낮은 이유를 구분하기 어렵다.

고객이 프로모션을 확인하지 않은 것인지, 확인했지만 조건을 달성하지 않은 것인지에 따라 개선 방법이 달라진다.

## 3. 핵심 문제

> 어떤 고객군·프로모션 조합에서 수신 후 열람 또는 열람 후 완료의 병목이 발생하며, 무엇을 우선 개선해야 하는가?

## 4. 대시보드가 지원할 의사결정

- 수신 후 열람이 낮은 고객군과 프로모션을 찾는다.
- 열람 후 완료가 낮은 고객군과 프로모션을 찾는다.
- 병목이 발생한 단계에 따라 개선 우선순위를 정한다.
- 성과가 좋은 고객군·프로모션 조합을 우선 발송 후보로 검토한다.

## 5. 원인 분해

### 고객 특성

- 성별
- 연령대
- 소득 구간
- 가입 기간
- 기존 결제 수준

### 프로모션 조건

- 프로모션 유형
- 보상금액
- 달성 조건
- 유효기간
- 발송 채널 조합

### 고객 반응 단계

- 프로모션 수신
- 프로모션 열람
- 프로모션 완료

## 6. 데이터상 제약사항

- 정보 제공형 프로모션은 완료 조건이 없어 열람까지만 평가해야 한다.
- 결제 기록에는 프로모션 ID가 없으므로 특정 결제를 프로모션에 직접 연결할 수 없다.
- 고객, 수신·열람 시점과 유효기간을 기준으로 프로모션 관련 결제를 추정할 수 있지만, 여러 프로모션이 겹친 경우 정확한 구분이 어렵다.
- 개별 채널의 실제 열람 여부가 없어 채널별 효과를 직접 비교하기 어렵다.
- 고객정보가 없는 2,175명은 고객군 분석 시 별도 처리가 필요하다.

## 7. 팀 논의 사항

- 핵심 문제를 ‘고객별 추천’과 ‘퍼널 병목 진단’ 중 어느 방향으로 설정할지
- 주요 성과지표를 열람률과 열람 후 완료율로 설정할지
- 정보 제공형 프로모션을 별도로 분석할지
- 고객군과 프로모션 조건을 어떤 기준으로 구분할지

## 가설 및 측정 기준 검토

### 1. 핵심 문제

> 프로모션 설계와 고객 특성에 따라 프로모션 완료율에 차이가 있는가?

`offer completed`는 일반적인 주문 완료가 아니라 프로모션의 결제 조건을 달성한 기록이다. 따라서 분석에서는 ‘주문 완료율’이 아닌 ‘프로모션 완료율’이라는 용어를 사용한다.

### 2. 프로모션 유형별 평가 기준

정보 제공형 프로모션은 보상과 달성 조건이 없어 완료 이벤트가 발생하지 않는다. 따라서 정보 제공형을 BOGO·할인형과 동일한 완료율로 비교하지 않는다.

- 전체 프로모션 유형: 수신 후 열람률 비교
- BOGO·할인형: 수신 후 완료율 비교
- 정보 제공형: 열람률 중심으로 별도 평가

이에 따라 프로모션 유형에 관한 가설을 다음과 같이 구분한다.

- 프로모션 유형에 따라 열람률에 차이가 있을 것이다.
- 할인형 프로모션은 BOGO보다 프로모션 완료율이 높을 것이다.

### 3. 주요 가설

#### 프로모션 설계

- 보상 수준이 높을수록 프로모션 완료율이 높을 것이다.
- 유효기간이 길수록 프로모션 완료율이 높을 것이다.
- 달성 난이도가 높을수록 프로모션 완료율이 낮을 것이다.
- 발송 채널 조합에 따라 열람률과 완료율에 차이가 있을 것이다.

#### 고객 특성

- 여성 고객은 남성 고객보다 프로모션 완료율이 높을 것이다.
- 소득이 높은 고객은 소득이 낮은 고객보다 프로모션 완료율이 높을 것이다.
- 연령이 높은 고객은 연령이 낮은 고객보다 프로모션 완료율이 낮을 것이다.

### 4. 구간 기준

가설 단계에서는 높음·낮음의 방향만 설정하고, 실제 구간 기준은 지표 설계 단계에서 확정한다.

현재 데이터에서 확인한 기준값은 다음과 같다.

- 보상 평균: 4.2
- 유효기간 평균: 6.5일
- 달성 난이도 평균: 7.7
- 118세를 제외한 고객 평균 연령: 약 54.39세
- 고객 소득 중앙값: 64,000달러
- 고객 소득 평균: 약 65,405달러

연령을 평균 기준으로 구분할 경우 54세 이하와 55세 이상으로 나눌 수 있다. 소득 8만 달러 이상을 고소득층으로 정의할 경우에는 팀이 설정한 별도의 구간 기준임을 명시한다.

### 5. 완료율 정의

프로모션 완료율은 다음과 같이 정의한다.

> BOGO·할인형 프로모션 수신 건 중 완료된 건의 비율

정보 제공형 프로모션 수신 건은 완료율 계산의 분모에서 제외한다. 동일 고객이 같은 프로모션을 여러 번 받을 수 있으므로, 고객·프로모션·수신 시점과 유효기간을 기준으로 수신과 완료를 연결하는 방법은 전처리 단계에서 확정한다.

### 6. 해석 시 주의사항

- 여러 발송 채널이 하나의 프로모션에 함께 포함되어 있어 개별 채널의 독립적인 효과를 확정하기 어렵다.
- 채널 분석 결과는 인과관계가 아니라 채널 포함 여부 또는 채널 조합과 성과의 연관성으로 해석한다.
- 고객정보가 누락되고 나이가 118세로 기록된 2,175명은 고객 특성별 분석에서 별도로 처리한다.
- 프로모션 설계는 10개뿐이므로 보상·기간·난이도·채널 각각의 독립적인 효과를 강하게 단정하지 않는다.
