# 지표 지도 그리기

## Funnel & North Star Framework

### 1. 분석 목적

고객이 프로모션을 수신한 뒤 열람하고 조건을 달성하기까지의 과정을 퍼널로 구성한다.

각 단계의 전환율을 확인하고, 프로모션 설계와 고객 특성에 따라 성과가 어떻게 달라지는지 분석한다.

---

## 2. North Star Metric

### 인센티브형 프로모션의 열람 후 완료율

> 열람 후 유효기간 내 완료된 BOGO·할인형 프로모션 수  
> ÷ 열람된 BOGO·할인형 프로모션 수 × 100

열람 후 완료율은 고객이 프로모션을 확인한 후 실제로 조건을 달성한 정도를 보여준다.

정보 제공형 프로모션은 보상과 달성 조건이 없어 완료 이벤트가 발생하지 않으므로 North Star Metric 계산에서 제외한다.

---

## 3. 프로모션 퍼널

| 퍼널 단계 | 지표 | 계산 기준 | 의미 |
|---|---|---|---|
| 수신 | 프로모션 수신 건수 | `offer received` 기록 수 | 고객에게 발송된 프로모션 규모 |
| 수신 → 열람 | 수신 후 열람률 | 열람된 프로모션 수 ÷ 수신된 프로모션 수 × 100 | 수신한 프로모션을 확인한 비율 |
| 열람 → 완료 | 열람 후 완료율 | 열람 후 완료된 인센티브형 프로모션 수 ÷ 열람된 인센티브형 프로모션 수 × 100 | 프로모션을 확인한 후 조건을 달성한 비율 |
| 수신 → 완료 | 프로모션 완료율 | 완료된 인센티브형 프로모션 수 ÷ 수신된 인센티브형 프로모션 수 × 100 | 수신한 프로모션 중 최종 완료된 비율 |
| 결제 | 결제금액 | `transaction`의 `amount` 합계 | 고객의 전체 구매 규모를 보여주는 보조 지표 |

인센티브형 프로모션은 BOGO와 할인형 프로모션을 의미한다.

## 5. 지표 계산 기준

### 수신 후 열람률

> 열람된 프로모션 수 ÷ 수신된 프로모션 수 × 100

BOGO·할인·정보 제공형을 모두 포함해 계산할 수 있다.

### 프로모션 완료율

> 완료된 인센티브형 프로모션 수 ÷ 수신된 인센티브형 프로모션 수 × 100

완료 조건이 있는 BOGO와 할인형만 포함한다.

### 열람 후 완료율

> 열람 후 완료된 인센티브형 프로모션 수 ÷ 열람된 인센티브형 프로모션 수 × 100

열람 시점보다 완료 시점이 늦고, 프로모션 유효기간 안에 완료된 경우만 포함한다.

## 7. 측정 시 주의사항

- 동일 고객이 같은 프로모션을 여러 번 받을 수 있으므로 고객·프로모션·수신 시점을 기준으로 각 프로모션 수신 건을 구분해야 한다.
- 열람과 완료는 해당 프로모션의 수신 이후이면서 유효기간 안에 발생한 경우만 연결해야 한다.
- 정보 제공형 프로모션은 완료 조건이 없으므로 열람률까지만 평가한다.
- 여러 채널이 하나의 프로모션에 함께 포함되어 있어 개별 채널의 독립적인 효과를 확정하기 어렵다.
- 채널 분석은 채널의 인과효과가 아니라 채널 포함 여부 또는 채널 조합과 성과의 연관성으로 해석한다.
- 고객정보가 누락되고 나이가 118세로 기록된 2,175명은 고객 특성별 분석에서 별도로 처리한다.
- 결제 기록에는 프로모션 ID가 없으므로 특정 결제를 특정 프로모션의 성과로 직접 귀속할 수 없다.
- 프로모션 설계는 10개뿐이므로 보상·기간·난이도·채널 각각의 독립적인 효과를 강하게 단정하지 않는다.

# 지표 계산

## 1. 데이터 불러오기

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("raw")

portfolio = pd.read_json(
    DATA_DIR / "portfolio.json",
    lines=True
)

profile = pd.read_json(
    DATA_DIR / "profile.json",
    lines=True
)

transcript = pd.read_json(
    DATA_DIR / "transcript.json",
    lines=True
)

print(portfolio.shape)
print(profile.shape)
print(transcript.shape)
```

## 2. 프로모션 ID 추출

```python
transcript_work = transcript.copy()

transcript_work["offer_id"] = transcript_work["value"].apply(
    lambda x: x.get("offer id", x.get("offer_id"))
)

display(
    transcript_work[
        ["person", "event", "offer_id", "time"]
    ].head()
)
```

```python
offer_id_check = (
    transcript_work
    .groupby("event")["offer_id"]
    .agg(
        전체_기록="size",
        ID_존재="count"
    )
)

display(offer_id_check)
```

### 추출 결과

- 프로모션 수신 76,277건, 열람 57,725건, 완료 33,579건에서 프로모션 ID가 모두 정상적으로 추출되었다.
- 결제 138,953건에는 원본 데이터상 프로모션 ID가 없으므로 `offer_id`가 결측치로 남는 것이 정상이다.
- 프로모션 관련 기록에서 ID가 누락된 경우는 확인되지 않았다.

## 3. 수신 건별 열람·완료 연결

```python
# 프로모션 유효기간을 시간 단위로 변환
duration_hours = (
    portfolio.set_index("id")["duration"] * 24
).to_dict()

# 결제를 제외한 프로모션 행동만 선택
offer_events = transcript_work.loc[
    transcript_work["event"] != "transaction",
    ["person", "event", "time", "offer_id"]
].copy()

# 같은 시간에는 수신 → 열람 → 완료 순서로 처리
event_order = {
    "offer received": 0,
    "offer viewed": 1,
    "offer completed": 2
}

offer_events["event_order"] = (
    offer_events["event"].map(event_order)
)

offer_events = offer_events.sort_values(
    ["person", "offer_id", "time", "event_order"]
)

offer_instances = []

for (person, offer_id), group in offer_events.groupby(
    ["person", "offer_id"],
    sort=False
):
    received_instances = []

    for row in group.itertuples(index=False):

        if row.event == "offer received":
            received_instances.append({
                "customer_id": person,
                "offer_id": offer_id,
                "received_time": row.time,
                "expiry_time": (
                    row.time + duration_hours[offer_id]
                ),
                "viewed_time": None,
                "completed_time": None
            })

        elif row.event == "offer viewed":
            candidates = [
                instance
                for instance in received_instances
                if (
                    instance["received_time"]
                    <= row.time
                    <= instance["expiry_time"]
                    and instance["viewed_time"] is None
                )
            ]

            if candidates:
                latest_instance = max(
                    candidates,
                    key=lambda x: x["received_time"]
                )
                latest_instance["viewed_time"] = row.time

        elif row.event == "offer completed":
            candidates = [
                instance
                for instance in received_instances
                if (
                    instance["received_time"]
                    <= row.time
                    <= instance["expiry_time"]
                    and instance["completed_time"] is None
                )
            ]

            if candidates:
                latest_instance = max(
                    candidates,
                    key=lambda x: x["received_time"]
                )
                latest_instance["completed_time"] = row.time

    offer_instances.extend(received_instances)

offer_funnel = pd.DataFrame(offer_instances)

print("생성된 프로모션 수신 건:", len(offer_funnel))
```

### 프로모션 수신 단위 생성 결과

고객별 프로모션 성과를 계산하기 위해 행동 기록을 프로모션 수신 건 단위로 재구성했다.

- 원본 프로모션 수신 기록: 76,277건
- 생성된 프로모션 수신 단위: 76,277건
- 누락되거나 추가된 수신 기록: 0건

각 수신 건에는 고객 ID, 프로모션 ID, 수신 시점과 유효기간을 부여했다. 이후 유효기간 안에 발생한 열람 및 완료 기록을 해당 수신 건에 연결한다.

동일 고객이 동일한 프로모션을 여러 번 받은 경우에는 행동 발생 시점을 기준으로 유효한 가장 최근 수신 건에 연결하는 잠정 기준을 사용했다. 최종 분석 전 연결되지 않은 열람·완료 기록과 중복 기록을 추가로 점검해야 한다.

## 4. 프로모션 정보 연결

```python
offer_funnel = offer_funnel.merge(
    portfolio[
        [
            "id",
            "offer_type",
            "reward",
            "difficulty",
            "duration",
            "channels"
        ]
    ],
    left_on="offer_id",
    right_on="id",
    how="left"
).drop(columns="id")

offer_funnel["viewed"] = (
    offer_funnel["viewed_time"].notna()
)

offer_funnel["completed"] = (
    offer_funnel["completed_time"].notna()
)

offer_funnel["viewed_before_completed"] = (
    offer_funnel["viewed"]
    & offer_funnel["completed"]
    & (
        offer_funnel["viewed_time"]
        <= offer_funnel["completed_time"]
    )
)

offer_funnel["completed_without_prior_view"] = (
    offer_funnel["completed"]
    & ~offer_funnel["viewed_before_completed"]
)

display(offer_funnel.head())
```

```python
print("[프로모션 정보 결측치]")
print(
    offer_funnel[
        [
            "offer_type",
            "reward",
            "difficulty",
            "duration",
            "channels"
        ]
    ].isna().sum()
)

print("\n[만료시간 계산 확인]")
expiry_check = (
    offer_funnel["expiry_time"]
    == offer_funnel["received_time"]
    + offer_funnel["duration"] * 24
)

print("만료시간이 잘못 계산된 건수:", (~expiry_check).sum())

print("\n[정보 제공형 완료 확인]")
informational_completed = (
    (offer_funnel["offer_type"] == "informational")
    & offer_funnel["completed"]
).sum()

print("정보 제공형 완료 건수:", informational_completed)
```

### 프로모션 정보 및 행동 순서 연결

 ### 확인 목적

 프로모션 수신 건에 유형, 보상, 달성 조건, 유효기간과 발송 채널을 연결하고, 열람과 완료가 발생한 순서를 구분했다.

 ### 생성된 주요 변수

 - `viewed`: 유효기간 안에 프로모션을 열람했는지
 - `completed`: 유효기간 안에 프로모션 조건을 달성했는지
 - `viewed_before_completed`: 프로모션을 열람한 뒤 완료했는지
 - `completed_without_prior_view`: 프로모션을 열람하기 전에 완료했는지

 ### 해석

 일부 고객은 프로모션을 열람하기 전에 완료 조건을 달성했다. 따라서 단순히 열람 기록과 완료 기록이 모두 존재하는지만 확인하면 안 되고, 
  열람과 완료의 시간 순서를 함께 고려해야 한다.

 정보 제공형 프로모션은 완료 조건이 없으므로 완료 여부가 아닌 열람 여부를 중심으로 평가한다.

## 5. 전체 퍼널 지표 계산

```python
incentive_mask = offer_funnel["offer_type"].isin(
    ["bogo", "discount"]
)

all_received = len(offer_funnel)
all_viewed = offer_funnel["viewed"].sum()

incentive_received = incentive_mask.sum()
incentive_viewed = (
    incentive_mask & offer_funnel["viewed"]
).sum()

incentive_completed = (
    incentive_mask & offer_funnel["completed"]
).sum()

viewed_then_completed = (
    incentive_mask
    & offer_funnel["viewed_before_completed"]
).sum()

completed_without_view = (
    incentive_mask
    & offer_funnel["completed_without_prior_view"]
).sum()

print("[전체 프로모션]")
print("수신 건수:", all_received)
print("열람 건수:", all_viewed)
print("수신 후 열람률:", round(all_viewed / all_received * 100, 2), "%")

print("\n[BOGO·할인형 프로모션]")
print("수신 건수:", incentive_received)
print("열람 건수:", incentive_viewed)
print("완료 건수:", incentive_completed)

print(
    "수신 후 열람률:",
    round(incentive_viewed / incentive_received * 100, 2),
    "%"
)

print(
    "수신 후 완료율:",
    round(incentive_completed / incentive_received * 100, 2),
    "%"
)

print(
    "열람 후 완료율:",
    round(viewed_then_completed / incentive_viewed * 100, 2),
    "%"
)

print(
    "완료 중 사전 열람이 없었던 비율:",
    round(completed_without_view / incentive_completed * 100, 2),
    "%"
)
```

### 7. 프로모션 유형별 퍼널 지표 계산

```python
import numpy as np

# 7. 프로모션 유형별 퍼널 지표 계산
offer_type_summary = (
    fully_observed
    .groupby("offer_type")
    .agg(
        수신_건수=("offer_id", "size"),
        열람_건수=("viewed", "sum"),
        완료_건수=("completed", "sum"),
        열람후_완료_건수=("viewed_before_completed", "sum"),
        사전열람없는_완료_건수=(
            "completed_without_prior_view",
            "sum"
        )
    )
)

offer_type_summary["수신후_열람률"] = (
    offer_type_summary["열람_건수"]
    / offer_type_summary["수신_건수"]
    * 100
)

# informational은 완료율 적용 대상이 아니므로 NaN 처리
incentive_type = offer_type_summary.index.isin(
    ["bogo", "discount"]
)

offer_type_summary["열람후_완료율"] = np.where(
    incentive_type,
    (
        offer_type_summary["열람후_완료_건수"]
        / offer_type_summary["열람_건수"]
        * 100
    ),
    np.nan
)

offer_type_summary["사전열람없는_완료_비율"] = np.where(
    incentive_type,
    (
        offer_type_summary["사전열람없는_완료_건수"]
        / offer_type_summary["완료_건수"]
        * 100
    ),
    np.nan
)

rate_columns = [
    "수신후_열람률",
    "열람후_완료율",
    "사전열람없는_완료_비율"
]

offer_type_summary[rate_columns] = (
    offer_type_summary[rate_columns].round(2)
)

display(offer_type_summary)
```

`social`이 포함된 프로모션은 대체로 높은 열람률을 보였지만, 채널·보상·난이도·기간이 함께 다르므로 개별 채널의 독립적인 효과로 단정하지 않는다.

### 9. 프로모션 설계 조건별 퍼널 지표

```python
# 9. 프로모션 설계 조건별 퍼널 지표

design_analysis = fully_observed.copy()

# 가설 트리에서 정한 평균 기준으로 구간 생성
design_analysis["reward_level"] = np.where(
    design_analysis["reward"] > 4.2,
    "평균 초과",
    "평균 이하"
)

design_analysis["duration_level"] = np.where(
    design_analysis["duration"] > 6.5,
    "평균 초과",
    "평균 이하"
)

design_analysis["difficulty_level"] = np.where(
    design_analysis["difficulty"] > 7.7,
    "평균 초과",
    "평균 이하"
)


def summarize_factor(data, column, factor_name):
    # 전체 프로모션의 열람률
    view_summary = (
        data.groupby(column)
        .agg(
            수신_건수=("offer_id", "size"),
            열람_건수=("viewed", "sum")
        )
    )

    view_summary["수신후_열람률"] = (
        view_summary["열람_건수"]
        / view_summary["수신_건수"]
        * 100
    )

    # BOGO·할인형 중 실제 열람된 건의 완료율
    incentive_viewed = data.loc[
        data["offer_type"].isin(["bogo", "discount"])
        & data["viewed"]
    ]

    completion_summary = (
        incentive_viewed.groupby(column)
        .agg(
            열람된_인센티브_건수=("offer_id", "size"),
            열람후_완료_건수=(
                "viewed_before_completed",
                "sum"
            )
        )
    )

    completion_summary["열람후_완료율"] = (
        completion_summary["열람후_완료_건수"]
        / completion_summary["열람된_인센티브_건수"]
        * 100
    )

    result = (
        view_summary
        .join(completion_summary, how="left")
        .reset_index()
        .rename(columns={column: "구분"})
    )

    result.insert(0, "설계요소", factor_name)

    rate_columns = [
        "수신후_열람률",
        "열람후_완료율"
    ]

    result[rate_columns] = (
        result[rate_columns].round(2)
    )

    return result


# 보상·기간·난이도
reward_summary = summarize_factor(
    design_analysis,
    "reward_level",
    "보상"
)

duration_summary = summarize_factor(
    design_analysis,
    "duration_level",
    "유효기간"
)

difficulty_summary = summarize_factor(
    design_analysis,
    "difficulty_level",
    "난이도"
)


# 채널 포함 여부
channel_results = []

for channel in ["web", "email", "mobile", "social"]:
    column_name = f"{channel}_included"

    design_analysis[column_name] = (
        design_analysis["channels"]
        .apply(
            lambda channel_list: (
                "포함"
                if channel in channel_list
                else "미포함"
            )
        )
    )

    channel_result = summarize_factor(
        design_analysis,
        column_name,
        channel
    )

    channel_results.append(channel_result)


# 하나의 표로 결합
design_summary = pd.concat(
    [
        reward_summary,
        duration_summary,
        difficulty_summary,
        *channel_results
    ],
    ignore_index=True
)

display(design_summary)
```

mobile과 social이 포함된 프로모션은 수신 후 열람률이 높았지만 열람 후 완료율은 더 높지 않았다. 따라서 채널 포함 여부는 주로 열람 단계와 연관된 것으로 보인다.

모든 프로모션에 email이 포함되어 있어 email 포함 여부에 따른 성과 차이는 비교할 수 없다.

각 프로모션은 유형, 보상, 기간, 난이도와 채널 조합이 함께 다르다. 따라서 결과는 설계 요소와 성과의 연관성으로 해석하고, 특정 요소의 독립적인 인과효과로 단정하지 않는다.

### 10. 고객 특성별 퍼널 지표

```python
# 10. 고객 특성별 퍼널 지표

customer_analysis = fully_observed.merge(
    profile.rename(columns={"id": "customer_id"}),
    on="customer_id",
    how="left",
    validate="many_to_one"
)

# 고객 구간 생성
customer_analysis["gender_group"] = (
    customer_analysis["gender"].fillna("정보 없음")
)

customer_analysis["income_group"] = np.where(
    customer_analysis["income"].isna(),
    "정보 없음",
    np.where(
        customer_analysis["income"] >= 80000,
        "8만 달러 이상",
        "8만 달러 미만"
    )
)

customer_analysis["age_group"] = np.where(
    customer_analysis["age"] == 118,
    "정보 없음",
    np.where(
        customer_analysis["age"] >= 55,
        "55세 이상",
        "54세 이하"
    )
)

gender_summary = summarize_factor(
    customer_analysis,
    "gender_group",
    "성별"
)

income_summary = summarize_factor(
    customer_analysis,
    "income_group",
    "소득"
)

age_summary = summarize_factor(
    customer_analysis,
    "age_group",
    "연령"
)

customer_summary = pd.concat(
    [
        gender_summary,
        income_summary,
        age_summary
    ],
    ignore_index=True
)

display(customer_summary)
```
